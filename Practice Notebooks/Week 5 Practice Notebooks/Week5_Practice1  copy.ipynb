````python

#
for item_number in range (1, len(values) + 1) :
  max_val = max(values)
  max_index = values.index(max_val)
  corresponding_weight = weights[max_index]

  print (f'Choose item number {item_number} with index {max_index}, value = {max_val} and weight = {corresponding_weight}')
  # set value to zero so we don't find it again
  values[max_index] = 0


````
